```python
# First let's import the packages we will use in this project
# You can do this all now or as you need them
import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt
import matplotlib.mlab as mlab
import matplotlib
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8)


# Now we need to read in the data
df = pd.read_csv(r'C:\Users\joshu\Downloads\archive\movies.csv')
```

```python
# Let's look at the data

df.head()
```

```python
# Dropped nulls

df = df.dropna()
```

```python

for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print("{} - {}%".format(col, pct_missing))
```

```python
# Let's look at our data types

print(df.dtypes)
```

```python
# updating the datatypes

df["budget"] = df["budget"].astype("int64")
df["gross"] = df["gross"].astype("int64")
```

```python
df.head()
```

```python
# make Year column correct


df['yearcorrect'] = df['released'].str.extract(pat = '([0-9]{4})').astype(int)
df.head()
```

```python
df = df.sort_values(by="gross", inplace=False, ascending=False)
```

```python
pd.set_option("display.max_rows", None)
```

```python
# Drop any duplicate

df["company"].drop_duplicates().sort_values(ascending=False)
```

```python
df.head()
```

```python
# Scatter plot showing budget vs. gross

plt.scatter(x=df["budget"], y=df["gross"])

plt.title("Budget vs Gross Earnings")
plt.xlabel("Budget for Film")
plt.ylabel("Gross Earnings")

plt.show()
```

```python
df.head()
```

```python
# Plot budget vs. gross using seaborn

sns.regplot(x="budget", y="gross", data=df, scatter_kws={"color": "red"}, line_kws={"color": "blue"})
```

```python
df.corr(numeric_only=True)
```

```python

correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(correlation_matrix, annot=True)

plt.title("Correlation Matrix for Numeric Features")
plt.xlabel("Movie Features")
plt.ylabel("Movie Features")

plt.show()

```

```python
# Some categories are not numbers

df.head()
```

```python
df_numerized = df.copy()

for col_name in df_numerized.columns:
    if (df_numerized[col_name].dtype == "str"):
        df_numerized[col_name] = df_numerized[col_name].astype("category")
        df_numerized[col_name] = df_numerized[col_name].cat.codes

df_numerized.head()
```

```python
df.head()
```

```python
correlation_matrix = df_numerized.corr(numeric_only=True)

sns.heatmap(correlation_matrix, annot=True)

plt.title("Correlation Matrix for Numeric Features")
plt.xlabel("Movie Features")
plt.ylabel("Movie Features")

plt.show()
```

```python
df_numerized.corr()
```

```python
correlation_mat = df_numerized.corr()

corr_pairs = correlation_mat.unstack()

corr_pairs
```

```python
sorted_pairs = corr_pairs.sort_values()

sorted_pairs
```

```python
high_corr = sorted_pairs[sorted_pairs > 0.5]

high_corr
```
